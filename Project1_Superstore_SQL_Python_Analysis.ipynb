```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
import sqlite3

conn = sqlite3.connect("Superstore.db")
```

```python
pd.read_sql("""
SELECT name 
FROM sqlite_master 
WHERE type='table';
""", conn)
```

```python
df.to_sql("superstore_orders", conn, if_exists="replace", index=False)
```

```python
pd.read_sql("""
SELECT name 
FROM sqlite_master 
WHERE type='table';
""", conn)
```

```python
pd.read_sql("SELECT * FROM superstore_orders LIMIT 5;", conn)
```

```python
df_kpi = pd.read_sql("""
SELECT 
    ROUND(SUM(sales),2) AS total_sales,
    ROUND(SUM(profit),2) AS total_profit
FROM superstore_orders;
""", conn)

df_kpi
```

```python
df_region = pd.read_sql("""
SELECT 
    region,
    ROUND(SUM(sales),2) AS sales,
    ROUND(SUM(profit),2) AS profit
FROM superstore_orders
GROUP BY region
ORDER BY sales DESC;
""", conn)

df_region
```

```python
df_category = pd.read_sql("""
SELECT 
    category,
    ROUND(SUM(sales),2) AS sales,
    ROUND(SUM(profit),2) AS profit
FROM superstore_orders
GROUP BY category;
""", conn)

df_category
```

```python
df_segment = pd.read_sql("""
SELECT 
    segment,
    ROUND(SUM(sales),2) AS sales,
    ROUND(SUM(profit),2) AS profit
FROM superstore_orders
GROUP BY segment;
""", conn)

df_segment
```

```python
df_discount = pd.read_sql("""
SELECT 
    discount,
    ROUND(SUM(sales),2) AS sales,
    ROUND(SUM(profit),2) AS profit
FROM superstore_orders
GROUP BY discount
ORDER BY discount;
""", conn)

df_discount
```

```python
df_kpi.to_csv("kpi_summary.csv", index=False)
df_region.to_csv("region_analysis.csv", index=False)
df_category.to_csv("category_analysis.csv", index=False)
df_segment.to_csv("segment_analysis.csv", index=False)
df_discount.to_csv("discount_analysis.csv", index=False)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(df_region['region'], df_region['sales'], label='Sales')
plt.bar(df_region['region'], df_region['profit'], label='Profit')

plt.title("Sales & Profit by Region")
plt.xlabel("Region")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.bar(df_category['category'], df_category['profit'])

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.plot(df_discount['discount'], df_discount['profit'], marker='o')

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.pie(
    df_segment['sales'],
    labels=df_segment['segment'],
    autopct='%1.1f%%'
)

plt.title("Sales Contribution by Segment")
plt.show()
```
